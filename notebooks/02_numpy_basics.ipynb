# 🔢 NumPy Fundamentals

```python
import numpy as np
```

## 1. Array Creation

```python
a = np.array([1, 2, 3])
b = np.zeros((3, 3))
c = np.ones((2, 2))
d = np.arange(10)
```

## 2. Array Operations

```python
print(a + 5)
print(a * 2)
print(a + np.array([3,2,1]))
```

## 3. Indexing & Slicing

```python
print(a[0])
print(d[2:5])
```

## 4. Linear Algebra

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(np.dot(A, B))
print(A @ B)
```
