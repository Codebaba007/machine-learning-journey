# 🐼 Pandas Data Analysis

```python
import pandas as pd
import numpy as np
```

## 1. Series & DataFrame

```python
s = pd.Series([1, 3, 5, np.nan, 6, 8])
df = pd.DataFrame({'A': 1., 'B': pd.Timestamp('20130102'), 'C': pd.Series(1, index=list(range(4)), dtype='float32')})
```

## 2. Data Selection

```python
print(df.loc[:, ['A', 'B']])
print(df.iloc[1:3, 0:2])
print(df[df['A'] > 0])
```

## 3. Data Cleaning

```python
df1 = df.reindex(index=list(range(4)), columns=list(df.columns) + ['E'])
df1.loc[0:1, 'E'] = 1
print(df1.dropna(how='any'))
print(df1.fillna(value=5))
```

## 4. GroupBy & Aggregation

```python
df2 = pd.DataFrame({'A': ['foo', 'bar', 'foo', 'bar'], 'B': ['one', 'one', 'two', 'three'], 'C': np.random.randn(4), 'D': np.random.randn(4)})
print(df2.groupby('A')[['C', 'D']].sum())
```
